```python
import pandas as pd
import matplotlib.pyplot as plt
```

# Read Data

```python
datapath = "/home/usc/ie/tam/nicf_first_semester/NiCf_analysis/teresa_analysis/"    #ahora los csv están en mi carpeta
filename_bkg = "df_bkgFromKaMing_forClaudeReco_noCuts.csv" 
data_bkg     = pd.read_csv(datapath+filename_bkg)

filename_sig = "df_sigFromKaMing_forClaudeReco_noCuts.csv"
data_sig     = pd.read_csv(datapath+filename_sig)
```

# nHits

```python
nhits_sig = data_sig.groupby("candidate_id")["hit_pmt_calibrated_times"].count()
nhits_bkg = data_bkg.groupby("candidate_id")["hit_pmt_calibrated_times"].count()
```

```python
plt.hist(nhits_sig, 100, histtype="step", label="SIG+BKG");
plt.hist(nhits_bkg, 100, histtype="step", label="BKG");
plt.xlabel("No. Hits Per Candidate [hits]");
# plt.yscale("log")
plt.legend();
```

# tRMS

```python
trms_sig = data_sig.groupby("candidate_id")["hit_pmt_calibrated_times"].std()
trms_bkg = data_bkg.groupby("candidate_id")["hit_pmt_calibrated_times"].std()
```

```python
plt.hist(trms_sig, 1000, histtype="step", label="SIG+BKG");
plt.hist(trms_bkg, 1000, histtype="step", label="BKG");
plt.xlabel("$t_{RMS}$ [ns]");
plt.legend();
```
